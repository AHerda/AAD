## Part i)

Wraz ze wzrostem poziomu szumu przedziały ufności dla β0 i β1 stają się szersze, wskazując na większą niepewność w oszacowaniach parametrów. Przy mniejszym szumie przedziały są węższe, co odzwierciedla bardziej precyzyjne oszacowania.

```python
noises = [0.1, 0.25, 0.5]

def conf_intervals(X, y, model):
    n = len(y)
    X_design = np.column_stack((np.ones(n), X.flatten()))
    y_pred = model.predict(X)
    residuals = y - y_pred
    s_squared = np.sum(residuals**2) / (n - 2)
    XTX_inv = np.linalg.inv(X_design.T @ X_design)
    se = np.sqrt(np.diag(s_squared * XTX_inv))
    beta_hat = np.array([model.intercept_, model.coef_[0]])
    t_val = stats.t.ppf(0.975, df=n-2)
    ci_lower = beta_hat - t_val * se
    ci_upper = beta_hat + t_val * se
    return (ci_lower, ci_upper)

Models = []
Ys = []
Eps = []

np.random.seed(1)
for noise in noises:
    eps_i = np.random.normal(0, noise, size=len(x))
    y_i = -1 + 0.5 * x + eps_i
    model_i = LinearRegression().fit(X, y_i)
    Models.append(model_i)
    Ys.append(y_i)
    Eps.append(eps_i)

for i, noise in enumerate(noises):
    model = Models[i]
    y = Ys[i]
    eps = Eps[i]
    ci = conf_intervals(X, y, model)
    print(f"Noise = {noise}")
    print(f"  β0 95% CI: [{ci[0][0]:.3f}, {ci[1][0]:.3f}]")
    print(f"  β1 95% CI: [{ci[0][1]:.3f}, {ci[1][1]:.3f}]")
    print()
```
